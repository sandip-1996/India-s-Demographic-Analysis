```python
import json
from collections import defaultdict
def checking_duplicates(column_metadata):
    grouped_duplicates = defaultdict(list)
    for keys, values in column_metadata.items():
        grouped_duplicates[values['name']].append(keys)

    duplicates = {val: key for val, key in grouped_duplicates.items() if len(key) > 1}
    print (duplicates)

def save_metadata(variables, output_dir):
    with open(output_dir / "metadata.json", "w", encoding="utf-8") as file:
       # json.dump(variables, file, indent=4)
       pass
# check the metadata.json file
```

```python
# metadata generator it create abbriviations of each column by concatenating first letter of every words of every columns.
import re
match = r'15|19|40|50|60|65|80'

data_table = [Base_data, Population_data, Fertility_data, Mortality_data, Migration_data]
data_name = ['Base_data', 'Population_data', 'Fertility_data', 'Mortality_data', 'Migration_data']
output_dir = Path.cwd().parent / "utils"
STOP_WORDS = {
    "of", "the", "a", "an",
    "as", "at", "to", "on", "in",
    "and", "or", "by", "for",
    "with", "per", "both",
    "before", "between", "under",
    "minus", "about"
}
STOP_NUMBERS = {'1', '000'}

all_metadata = {}
for data_,name_ in zip(data_table,data_name):
    col_umn = data_.columns.to_list()

    column_metadata = {}
    for col in col_umn:
        # checking first 3 words and extract first letter of word and make separate string
        split_text = re.findall(r"[^()]+", col)
        word_ = re.findall(r"[a-zA-Z0-9]+", split_text[0])
        word_ = [word for word in word_ if word.lower() not in STOP_WORDS]

        string_T = ""
        if len(word_) == 1:
            string_T = str(word_[0]).upper()
        else:
            for i in range(0,3):
                if 0 <= i < len(word_):
                    letter_i = list(word_[i].upper())
                    string_T += f"{letter_i[0]}"
                else:
                    pass

        # checking integer and make seperate string
        string_N = ""
        numbers =  re.findall(r'\d+', split_text[0])
        numbers = [number for number in numbers if number not in STOP_NUMBERS]

        if len(numbers) == 1:
            string_N = f'{numbers[0]}'
        if len(numbers) > 1 and len(numbers) <3:
            if numbers[0] != numbers[1]:
                string_N = f'{numbers[0]}_{numbers[1]}'
        if len(numbers) == 0:
            string_N = ""

        # concatenate two string
        string = string_T+"_"+string_N

        column_metadata[col] = {
                "name": string,
                "unit": split_text[-1]
            }
        string = ""
        # print(column_metadata)
    checking_duplicates(column_metadata)

    all_metadata[name_] = column_metadata
    save_metadata(all_metadata, output_dir)
```

```python
def SQL_structure(table_name, data_table):
    # make SQL column structure and save it on SQL folder.
    dtype_map = {
    'str': 'VARCHAR(64)',  
    'float64': 'DOUBLE PRECISION', # 64 bit floating point data
    'int64': 'BIGINT' # 8 byte (64 bit) int
    }

    output_dir = Path.cwd().parent / "SQL"

    for table,data_ in zip(table_name, data_table):
        sql = f"CREATE TABLE {table} (\n"
        for col, dtype in data_.dtypes.items():
            sql += f'   {col} {dtype_map.get(str(dtype), "TEXT")},\n' # get(key, default value)
            
        sql = sql.rstrip(",\n") + "\n" # right strip remove trailing ch from right side
        sql = sql + ")"

        with open(output_dir / f"{table}.sql", 'w') as f:
            f.write(sql)
```
